# Importation des packages

```python
import re
```

# Créer des fichiers

```python
test_list = ["tomate", "poire", "pomme", "cerise", "Fraise", "fraise", "courgette", "POMME3", "afraise"]
```

# Créer des regex

# Exercice 1

Mots qui contient 'se' est en fin de nom

```python
# Regex pour vérifier si le mot se termine par "se"
pattern = r'\b\w*se\b'

# Utiliser une liste en compréhension pour filtrer les mots qui correspondent à la regex
matches = [word for word in test_list if re.search(pattern, word)]

print(matches)
```

Réponse : ['cerise', 'Fraise', 'fraise', 'afraise']

# Exercice 2

Mots qui ont deux lettres qui se suivent

```python
pattern = r'\b\w*(.)\1\w*\b'

# Utiliser une liste en compréhension pour filtrer les mots qui correspondent à la regex
matches = [word for word in test_list if re.search(pattern, word)]

print(matches)
```

Réponse : ['pomme', 'courgette', 'POMME3']

# Exercice 3

Mots qui contiennent un chiffre

```python
pattern = r'\b\w*\d\w*\b'

# Utiliser une liste en compréhension pour filtrer les mots qui correspondent à la regex
matches = [word for word in test_list if re.search(pattern, word)]

print(matches)
```

Réponse : ['POMME3']

# Exercice 4

Trouver tous les mots qui contiennent au moins une lettre en majuscule

```python
pattern = r'\w*[A-Z]\w*'

# Utiliser une liste en compréhension pour filtrer les mots qui correspondent à la regex
matches = [word for word in test_list if re.search(pattern, word)]

print(matches)
```

# Exercice 8
